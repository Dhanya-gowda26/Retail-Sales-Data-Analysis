```python
import pandas as pd

url = "https://raw.githubusercontent.com/selva86/datasets/master/supermarket_sales.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
sales_by_product = df.groupby("Product line")["Total"].sum().sort_values(ascending=False)

sales_by_product
```

```python
import matplotlib.pyplot as plt

sales_by_product.plot(kind="bar", figsize=(10, 5))

plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sales_by_city = df.groupby("City")["Total"].sum().sort_values(ascending=False)

sales_by_city
```

```python
sales_by_city.plot(kind="bar", figsize=(8, 5))

plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
payment_counts = df["Payment"].value_counts()

payment_counts
```

```python
payment_counts.plot(kind="bar", figsize=(8, 5))

plt.title("Payment Method Distribution")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
avg_sales_by_product = df.groupby("Product line")["Total"].mean().sort_values(ascending=False)

avg_sales_by_product
```

```python
avg_rating = df.groupby("Product line")["Rating"].mean().sort_values(ascending=False)

avg_rating
```

```python
avg_rating.plot(kind="bar", figsize=(10, 5))

plt.title("Average Customer Rating by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sales_by_gender = df.groupby("Gender")["Total"].sum().sort_values(ascending=False)

sales_by_gender
```

```python
sales_by_gender.plot(kind="bar", figsize=(7, 5))

plt.title("Total Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
sales_by_customer_type = df.groupby("Customer type")["Total"].sum().sort_values(ascending=False)

sales_by_customer_type
```

```python
sales_by_customer_type.plot(kind="bar", figsize=(7, 5))

plt.title("Total Sales by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
total_sales = df["Total"].sum()
total_sales
```

```python
df["Date"] = pd.to_datetime(df["Date"])

monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total"].sum()

monthly_sales
```

```python
monthly_sales.plot(kind="line", marker="o", figsize=(8, 5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
correlation = df[[
    "Unit price",
    "Quantity",
    "Tax 5%",
    "Total",
    "cogs",
    "gross income",
    "Rating"
]].corr()

correlation
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## Key Insights

1. Food and beverages generated the highest total sales among all product lines.
2. Health and beauty had the lowest total sales.
3. Naypyitaw recorded the highest total sales among the three cities.
4. E-wallet was the most commonly used payment method, followed closely by cash.
5. Food and beverages received the highest average customer rating.
6. Female customers generated slightly higher total sales than male customers.
7. Member customers generated higher total sales than normal customers.
8. January recorded the highest monthly sales, while February had the lowest.
9. Quantity and total sales showed a strong positive relationship.
10. The overall average transaction value was approximately 322,966.75.
